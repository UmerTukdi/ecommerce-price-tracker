```python
import pandas as pd
import matplotlib.pyplot as plt

# Load scraped data
df = pd.read_csv("data/products.csv")


df.head()
```

```python
df.info()
df.describe()
```

```python
kpi_df = pd.DataFrame({
    "Metric": [
        "Total Products",
        "Average Price",
        "Minimum Price",
        "Maximum Price"
    ],
    "Value": [
        df.shape[0],
        round(df["Price"].mean(), 2),
        df["Price"].min(),
        df["Price"].max()
    ]
})

kpi_df
```

```python
**Insight:**
The dataset contains a limited number of products with a consistent pricing
range, making it suitable for comparative pricing analysis.
```

```python
plt.figure(figsize=(7,4))
plt.hist(df["Price"], bins=10)

plt.title("Distribution of Book Prices")
plt.xlabel("Price (£)")
plt.ylabel("Number of Books")
plt.grid(alpha=0.3)
plt.show()
```

```python
**Insight:**
Most products fall within a similar price range, indicating competitive and
standardized pricing across the platform.
```

```python
rating_counts = df["Rating"].value_counts()

plt.figure(figsize=(6,4))
rating_counts.plot(kind="bar")

plt.title("Distribution of Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Products")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
**Insight:**
A majority of products have higher ratings, suggesting overall positive customer
feedback and product quality.
```

```python
avg_price_by_rating = df.groupby("Rating")["Price"].mean()

plt.figure(figsize=(7,4))
avg_price_by_rating.plot(kind="bar")

plt.title("Average Price by Product Rating")
plt.xlabel("Rating")
plt.ylabel("Average Price (£)")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
**Insight:**
Average prices remain relatively similar across ratings, showing that higher
ratings do not necessarily correspond to higher prices.
```

```python
## Executive Summary
This analysis uses scraped e-commerce product data to examine pricing and
rating patterns. The insights reveal competitive pricing ranges, strong overall
ratings, and minimal relationship between price and perceived quality. Such
analysis can support pricing strategy, competitive benchmarking, and product
positioning decisions.
```
